```python
import pandas as pd
import json
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def get_info(df: pd.DataFrame, columns: list, perc_min=0.25, perc_max=0.75, verbose=False, print_boxplot=False):
    info_dict = {}
    for col in columns:
        min_val = df[col].min()
        max_val = df[col].max()
        mean_val = df[col].mean()
        q_min = df[col].quantile(perc_min)
        q_max = df[col].quantile(perc_max)
        iqr = q_max - q_min
        lower = q_min - 1.5*iqr
        upper = q_max + 1.5*iqr
        count_low = (df[col] < lower).sum()
        count_high = (df[col] > upper).sum()
        info_dict[col] = {}
        info_dict['min_val'] = min_val,
        info_dict['max_val'] = max_val,
        info_dict['mean_val'] = mean_val,
        info_dict['perc_min'] = perc_min,
        info_dict['perc_max'] = perc_max,
        info_dict['q_min'] = q_min,
        info_dict['q_max'] = q_max,
        info_dict['count_low'] = count_low,
        info_dict['count_high'] = count_high           
        if verbose:
            print('v'*50)
            print(f'Статистика по признаку {col}')
            print(f'Минимальное значение {min_val}')
            print(f'Максимальное значение {max_val}')
            print(f'Среднее значение {mean_val}')
            print(f'Значение соответствующее перцентилю {perc_min} - {q_min}')
            print(f'Значение соответствующее перцентилю {perc_max} - {q_max}')
            print(f'Количество выбросов ниже {perc_min} - {count_low}')
            print(f'Количество выбросов выше {perc_max} - {count_high}')
        if print_boxplot:
            plt.figure(figsize=(6, 4))
            ax = sns.boxplot(data=df, y=col)
            ax.set_title(f'{col}: min={min_val}, max={max_val}, mean={mean_val:.2f}')
            text = f'Количество записей меньше {lower} = {count_low} \n Количество значений больше {upper} = {count_high}'
            ax.text(1,
                    0,
                    text,
                    ha='right', 
                    va='center', 
                    fontsize=8, 
                    color='orange')
            ax.set_xlabel(f'Признак {col}')
            ax.set_ylabel('Значения')
            plt.show()
    return info_dict
```
